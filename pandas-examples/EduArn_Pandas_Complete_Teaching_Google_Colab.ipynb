```python
# 1. IMPORT PANDAS
import pandas as pd
import numpy as np

print('Pandas version:', pd.__version__)
print('Pandas is ready!')
```

## 2. SERIES

A Series is one-dimensional labeled data. Think of it as one column with an index.

```python
# Create a Series
sales = pd.Series([12000, 18500, 15000, 22000], name='Sales')
print(sales)
```

```python
# Series with custom labels
sales = pd.Series([12000, 18500, 15000, 22000], index=['Jan','Feb','Mar','Apr'], name='Sales')
print(sales)

print('Index:', sales.index)
print('Feb:', sales['Feb'])
print('First value:', sales.iloc[0])
```

### loc vs iloc

`loc` selects by label; `iloc` selects by integer position. citeturn0search8

```python
print('loc:', sales.loc['Mar'])
print('iloc:', sales.iloc[2])
print('First two:\n', sales.iloc[:2])
```

## 3. DATAFRAME

A DataFrame is a two-dimensional table of rows and columns. It is the main pandas structure for tabular data. citeturn0search0

```python
df = pd.DataFrame({
    'Employee':['Asha','Rahul','Priya','Arun','Neha','Vikram','Meena','Kiran'],
    'Department':['Sales','IT','Sales','HR','IT','Sales','HR','IT'],
    'City':['Hyderabad','Bengaluru','Hyderabad','Chennai','Bengaluru','Pune','Chennai','Hyderabad'],
    'Experience':[3,6,4,8,5,7,2,9],
    'Sales':[120000,180000,150000,90000,210000,195000,85000,230000],
    'Rating':[4.2,4.7,4.5,3.8,4.9,4.4,3.9,4.8]
})
display(df)
```

```python
print('Columns:', df.columns.tolist())
print('\nFirst row:')
display(df.iloc[0])
print('Rows:', len(df))
print('Columns:', len(df.columns))
```

## 5. RETRIEVE DATA

`[]` selects columns; `loc` uses labels/conditions; `iloc` uses positions.

```python
display(df['Employee'])
display(df[['Employee','Department','Sales']])
```

```python
# loc: rows and named columns
display(df.loc[0:2, ['Employee','Sales']])

# iloc: row/column positions
display(df.iloc[0:3, 0:3])
```

## 6. FILTER DATA

Think of filtering as an SQL `WHERE` condition.

```python
print('Sales > 150000')
display(df[df['Sales'] > 150000])

print('IT and rating >= 4.8')
display(df[(df['Department']=='IT') & (df['Rating']>=4.8)])

print('Hyderabad OR Bengaluru')
display(df[(df['City']=='Hyderabad') | (df['City']=='Bengaluru')])
```

## 7. MANIPULATE DATA

```python
# Add a column
df['Bonus'] = df['Sales'] * 0.10
display(df[['Employee','Sales','Bonus']])

# Modify it
df['Bonus'] = df['Sales'] * 0.12

# Rename
df = df.rename(columns={'Rating':'Performance_Rating'})

# Delete
df = df.drop(columns=['Bonus'])

display(df.head())
```

## 8. MISSING DATA

Pandas provides `isna()`, `fillna()` and `dropna()` for missing values. citeturn0search11turn0search4

```python
df_missing = df.copy()
df_missing.loc[2,'Performance_Rating'] = np.nan
df_missing.loc[5,'City'] = np.nan

display(df_missing)
print('Missing values by column:')
print(df_missing.isna().sum())
```

```python
# Fill numeric missing value with mean
df_missing['Performance_Rating'] = df_missing['Performance_Rating'].fillna(df_missing['Performance_Rating'].mean())
# Fill text missing value
df_missing['City'] = df_missing['City'].fillna('Unknown')
display(df_missing)
```

```python
# Drop rows containing missing values
df_test = df_missing.copy()
df_test.loc[0,'City'] = np.nan
print('Before:', len(df_test), 'rows')
print('After dropna:', len(df_test.dropna()), 'rows')
```

## 9. DUPLICATES

```python
df_dup = pd.concat([df, df.iloc[[0]]], ignore_index=True)
print('Rows:', len(df_dup))
print('Duplicate rows:', df_dup.duplicated().sum())
display(df_dup[df_dup.duplicated(keep=False)])

df_dup_clean = df_dup.drop_duplicates()
print('Rows after removing duplicates:', len(df_dup_clean))
```

## 10. SORTING

```python
# Highest sales first
display(df.sort_values('Sales', ascending=False)[['Employee','Sales']])

# Sort by Department, then Sales descending
display(df.sort_values(['Department','Sales'], ascending=[True,False])[['Employee','Department','Sales']])
```

## 11. STATISTICS

Pandas supports summary statistics such as mean, median, min, max, standard deviation and counts. citeturn0search0

```python
print('Total Sales:', df['Sales'].sum())
print('Average Sales:', df['Sales'].mean())
print('Median Sales:', df['Sales'].median())
print('Minimum Sales:', df['Sales'].min())
print('Maximum Sales:', df['Sales'].max())
print('Std Dev:', df['Sales'].std())
print('Employee Count:', df['Employee'].count())
```

```python
df[['Experience','Sales','Performance_Rating']].describe()
```

## 12. GROUPBY

GroupBy follows **split → apply → combine**: split records into groups, calculate something for each group, then combine the results. citeturn0search6

```python
print('Average sales by department:')
display(df.groupby('Department')['Sales'].mean().sort_values(ascending=False))

print('Department summary:')
summary = df.groupby('Department').agg(
    Employee_Count=('Employee','count'),
    Total_Sales=('Sales','sum'),
    Average_Sales=('Sales','mean'),
    Average_Rating=('Performance_Rating','mean')
).round(2)
display(summary)
```

```python
# Group by two columns
display(df.groupby(['City','Department'])['Sales'].sum().sort_values(ascending=False))
```

# 13. MINI PROJECT — Employee Sales Analysis

### Business questions
1. Who generated the highest sales?
2. Which department generated the most revenue?
3. What is the average sales value?
4. Which employees are high performers?
5. What is the average rating by department?

```python
# 1. Top 3 employees
display(df.nlargest(3,'Sales')[['Employee','Department','Sales','Performance_Rating']])

# 2. Department revenue
department_revenue = df.groupby('Department', as_index=False)['Sales'].sum().sort_values('Sales', ascending=False)
display(department_revenue)

# 3. Overall KPIs
print(f"Total Sales: ₹{df['Sales'].sum():,.0f}")
print(f"Average Sales: ₹{df['Sales'].mean():,.0f}")
print(f"Average Rating: {df['Performance_Rating'].mean():.2f}")
print('Top Salesperson:', df.loc[df['Sales'].idxmax(),'Employee'])

# 4. High performers
high = df[(df['Sales']>=180000) & (df['Performance_Rating']>=4.5)]
display(high[['Employee','Department','Sales','Performance_Rating']])
```

```python
# 5. Final department dashboard
dashboard = df.groupby('Department').agg(
    Employees=('Employee','count'),
    Total_Sales=('Sales','sum'),
    Avg_Sales=('Sales','mean'),
    Avg_Rating=('Performance_Rating','mean')
).round(2).sort_values('Total_Sales', ascending=False)
display(dashboard)
```
