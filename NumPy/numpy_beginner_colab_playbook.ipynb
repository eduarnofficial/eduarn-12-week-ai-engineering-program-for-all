## 1. Import NumPy

`np` is the conventional short name used for NumPy.

```python
import numpy as np

print('NumPy version:', np.__version__)
```

## 2. Create Your First Array

```python
numbers = np.array([10, 20, 30, 40, 50])

print(numbers)
print(type(numbers))
```

## 3. Basic Calculations

NumPy can perform an operation on every element of an array.

```python
numbers = np.array([10, 20, 30, 40, 50])

print('Add 10:', numbers + 10)
print('Multiply by 2:', numbers * 2)
print('Divide by 10:', numbers / 10)
```

## 4. Indexing

Remember: Python indexing starts at **0**.

```python
numbers = np.array([10, 20, 30, 40, 50])

print('First:', numbers[0])
print('Third:', numbers[2])
print('Last:', numbers[-1])
```

## 5. Slicing

```python
numbers = np.array([10, 20, 30, 40, 50])

print('First 3:', numbers[:3])
print('From index 2:', numbers[2:])
print('Index 1 to 3:', numbers[1:4])
print('Every second element:', numbers[::2])
```

## 6. Statistics

```python
marks = np.array([65, 72, 88, 91, 55])

print('Total:', np.sum(marks))
print('Average:', np.mean(marks))
print('Highest:', np.max(marks))
print('Lowest:', np.min(marks))
print('Standard deviation:', np.std(marks))
```

## 7. 2D Arrays

A 2D NumPy array can represent a table of data.

```python
data = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print(data)
```

```python
print('First row:', data[0])
print('First column:', data[:, 0])
print('Shape:', data.shape)
print('Dimensions:', data.ndim)
print('Total elements:', data.size)
```

## 8. Reshaping

```python
numbers = np.array([1, 2, 3, 4, 5, 6])

matrix = numbers.reshape(2, 3)

print(matrix)
```

## 9. Filtering Data

Boolean filtering is one of the most useful NumPy features for data analysis.

```python
marks = np.array([35, 67, 89, 42, 91, 28])

passed = marks[marks >= 40]
failed = marks[marks < 40]

print('Passed:', passed)
print('Failed:', failed)
```

## 10. Creating Sequences with arange()

```python
print(np.arange(1, 11))
print(np.arange(0, 20, 2))
```

## 11. Creating Evenly Spaced Values with linspace()

```python
numbers = np.linspace(0, 10, 5)

print(numbers)
```

## 12. Random Numbers

```python
np.random.seed(42)

marks = np.random.randint(0, 101, 10)

print('Random marks:', marks)
print('Average:', np.mean(marks))
```

# 🎯 Mini Project: Student Marks Analyzer

We will analyze a group of student marks.

```python
marks = np.array([78, 65, 92, 45, 88, 32, 74, 90, 56, 81])

print('Marks:', marks)
print('Total:', np.sum(marks))
print('Average:', np.mean(marks))
print('Highest:', np.max(marks))
print('Lowest:', np.min(marks))

passed = marks[marks >= 40]
failed = marks[marks < 40]

print('Passed:', passed)
print('Failed:', failed)
print('Number passed:', len(passed))
print('Number failed:', len(failed))
```
